# ✅Сравним нашу реализацию Гауссовского наивного Байеса с реализацией из Scikit-Learn

```python
from sklearn.datasets import load_iris
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

from bayes import BayesGaussian
from utils import decision_boundary_plot
```

# ✅Загрузим данные

```python
X1, y1 = load_iris(return_X_y=True, as_frame=True)
X1_train, X1_test, y1_train, y1_test = train_test_split(X1.values, y1.values, random_state=0)
print(X1, y1, sep='\n')
```

# ✅ Обучим самописный классификатор

```python
nb_clf = BayesGaussian()
nb_clf.fit(X1_train, y1_train)
nb_clf_pred_res = nb_clf.predict(X1_test)
nb_clf_accuracy = accuracy_score(y1_test, nb_clf_pred_res)

print(f'Naive Bayes classifier accucacy: {nb_clf_accuracy}')
```

```python
feature_indexes = [2, 3]
title1 = 'Bayes Gaussian CLF'
decision_boundary_plot(X1, y1, X1_train, y1_train, nb_clf, feature_indexes, title1)
```

# ✅Обучим классификатор из Scikit-learn

```python
sk_nb_clf = GaussianNB()
sk_nb_clf.fit(X1_train, y1_train)
sk_nb_clf_pred_res = sk_nb_clf.predict(X1_test)
sk_nb_clf_accuracy = accuracy_score(y1_test, sk_nb_clf_pred_res)

print(f'sk Naive Bayes classifier accucacy: {sk_nb_clf_accuracy}')
```

```python
feature_indexes = [2, 3]
title2 = 'Bayes CLF Scikit-Learn'
decision_boundary_plot(X1, y1, X1_train, y1_train, sk_nb_clf, feature_indexes, title2)
```
